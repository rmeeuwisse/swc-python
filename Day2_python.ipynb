## Analyze data from multiple files

```python
import glob
```

### Use glob to find all inflammation data files

```python
file_pattern = "python-novice-inflammation-data/data/inflam*.csv"
all_files = sorted(glob.glob(file_pattern))

print(all_files)
```

```python
data_arrays = []

for file in all_files:
    data = np.loadtxt(file, delimiter = ",")
    data_arrays.append(data)
    print("data from file:", file, "DONE")

```

```python
len(data_arrays)
data_arrays[0]

for array in data_arrays:
    # mean/max/min inflammation per day per file
    mean_perday = np.mean(array, axis = 0)
    max_perday = np.max(array, axis = 0)
    min_perday = np.min(array, axis = 0)

    # Create the figure of three subplots

    fig = plt.figure(figsize=(10,3))

    axes1 = fig.add_subplot(1,3,1)
    axes2 = fig.add_subplot(1,3,2)
    axes3 = fig.add_subplot(1,3,3)

    axes1.plot(mean_perday)
    axes2.plot(max_perday)
    axes3.plot(min_perday)

    axes1.set_ylabel("Mean")
    axes2.set_ylabel("Maxmimum")
    axes3.set_ylabel("Minimum")

    for ax in [axes1,axes2,axes3]:
        ax.set_xlabel("Day")
    
    fig.tight_layout()





```

```python
fig = plt.figure(figsize = (10,3))
plt.imshow(data_arrays[2])
plt.colorbar()

plt.title("Inflammation dataset 3")
plt.xlabel("Day")
plt.ylabel("Patient")
```
